### **Analysis**


 ( Hint: Group by, conditional filtering, Visualization )

##### Which membership type shows the highest total spending in the dataset?  (2 Points)

```python
# Group by 'Membership Type' and calculate total spending
total_spending_by_membership = ecommerce_data.groupby('Membership Type')['Total Spend'].sum().reset_index()

# Sort to find the membership type with the highest spending
highest_spending_membership = total_spending_by_membership.sort_values(by='Total Spend', ascending=False)

# Display the result
print("Total spending by membership type:")
print(highest_spending_membership)
```

```python
import matplotlib.pyplot as plt

# Plot total spending by membership type
plt.figure(figsize=(8, 6))
plt.bar(highest_spending_membership['Membership Type'], highest_spending_membership['Total Spend'], color='skyblue')
plt.xlabel('Membership Type')
plt.ylabel('Total Spending')
plt.title('Total Spending by Membership Type')
plt.xticks(rotation=45)
plt.show()
```

#### How does customer satisfaction impact total spending across different membership types? Which membership type spends more based on satisfaction levels? (2 Points)

```python
# Group by 'Membership Type' and 'Satisfaction Level' and calculate the average total spending
satisfaction_spending = ecommerce_data.groupby(['Membership Type', 'Satisfaction Level'])['Total Spend'].mean().reset_index()

# Sort the results for better readability
satisfaction_spending_sorted = satisfaction_spending.sort_values(by='Total Spend', ascending=False)

# Display the result
print("Average Total Spending based on Membership Type and Satisfaction Level:")
print(satisfaction_spending_sorted)
```

```python
# Determine which membership type spends more based on satisfaction levels:
#After analyzing the data, you can identify which membership type has the highest average total spending at each satisfaction level based on the printed results.

#Summary:
#The first code block will give you the average total spending for each combination of membership type and satisfaction level.
```

#### What is the average rating for customers who applied discounts, and does the membership type have any effect on the ratings? (2 points)

```python
# Filter customers who applied discounts
discounted_customers = ecommerce_data[ecommerce_data['Discount Applied'] == True]

# Calculate the average rating for these customers
average_rating_discounted = discounted_customers['Average Rating'].mean()

print(f"Average Rating for Customers Who Applied Discounts: {average_rating_discounted:.2f}")
```

```python
# Group by 'Membership Type' and calculate the average rating
average_rating_by_membership = ecommerce_data.groupby('Membership Type')['Average Rating'].mean().reset_index()

# Sort the results for better readability
average_rating_by_membership_sorted = average_rating_by_membership.sort_values(by='Average Rating', ascending=False)

# Display the results
print("Average Rating by Membership Type:")
print(average_rating_by_membership_sorted)
```

```python
# Plot average rating by membership type
plt.figure(figsize=(8, 6))
plt.bar(average_rating_by_membership_sorted['Membership Type'], average_rating_by_membership_sorted['Average Rating'], color='lightgreen')
plt.xlabel('Membership Type')
plt.ylabel('Average Rating')
plt.title('Average Rating by Membership Type')
plt.xticks(rotation=45)
plt.show()
```

```python
# Summary:
# The first code block calculates and displays the average rating for customers who applied discounts.
# The second block calculates the average rating by membership type, and the final visualization helps compare the ratings visually.
```

#### Analyze the Impact of Days Since Last Purchase on Satisfaction Level (2 Points)

```python
# Check the data types of the columns
print(ecommerce_data.dtypes)
```

```python
# Display unique values in the Satisfaction Level column
print(ecommerce_data['Satisfaction Level'].unique())
```

```python
# Map satisfaction levels to numeric values
satisfaction_mapping = {
    'Satisfied': 1,
    'Neutral': 0,
    'Dissatisfied': -1
}

# Convert the Satisfaction Level column to numeric
ecommerce_data['Satisfaction Level'] = ecommerce_data['Satisfaction Level'].map(satisfaction_mapping)

# Check if the conversion was successful
print(ecommerce_data['Satisfaction Level'].unique())
```

```python
# Group by 'Days Since Last Purchase' and calculate the average satisfaction level
satisfaction_by_days = ecommerce_data.groupby('Days Since Last Purchase')['Satisfaction Level'].mean().reset_index()

# Sort the results for better readability
satisfaction_by_days_sorted = satisfaction_by_days.sort_values(by='Days Since Last Purchase')

# Display the results
print("Average Satisfaction Level by Days Since Last Purchase:")
print(satisfaction_by_days_sorted)
```

```python
import matplotlib.pyplot as plt

# Plotting the average satisfaction level by days since last purchase
plt.figure(figsize=(10, 6))
plt.plot(satisfaction_by_days_sorted['Days Since Last Purchase'], satisfaction_by_days_sorted['Satisfaction Level'], marker='o', color='blue')
plt.xlabel('Days Since Last Purchase')
plt.ylabel('Average Satisfaction Level')
plt.title('Impact of Days Since Last Purchase on Satisfaction Level')
plt.grid()
plt.show()
```
